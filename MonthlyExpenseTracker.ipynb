```python
import csv
import os
from datetime import datetime

FILE = "expenses.csv"
expenses = []
monthly_budget = 0.0

def load_expenses():
    global expenses
    if not os.path.exists(FILE):
        return
    with open(FILE, "r", newline="") as f:
        reader = csv.DictReader(f)
        for row in reader:
            try:
                row["amount"] = float(row["amount"])
                if all([row.get("date"), row.get("category"), row.get("amount"), row.get("description")]):
                    expenses.append(row)
            except:
                continue

def save_expenses():
    with open(FILE, "w", newline="") as f:
        fieldnames = ["date", "category", "amount", "description"]
        writer = csv.DictWriter(f, fieldnames=fieldnames)
        writer.writeheader()
        writer.writerows(expenses)
    print("Expenses saved to file.")

def add_expense():
    date = input("Date YYYY-MM-DD: ")
    try:
        datetime.strptime(date, "%Y-%m-%d")
    except:
        print("Error: Invalid date format. Use YYYY-MM-DD")
        return
    
    category = input("Category - Food/Travel/etc: ").strip()
    if not category:
        print("Error: Category is required")
        return
    
    try:
        amount = float(input("Amount spent: "))
        if amount <= 0:
            print("Error: Amount must be > 0")
            return
    except:
        print("Error: Invalid amount")
        return
    
    description = input("Brief description: ").strip()
    if not description:
        print("Error: Description is required")
        return

    expense = {
        "date": date,
        "category": category,
        "amount": amount,
        "description": description
    }
    expenses.append(expense)
    print("Expense added!")

def view_expenses():
    if not expenses:
        print("No expenses recorded yet.")
        return
    
    print("\n--- All Expenses ---")
    for i, e in enumerate(expenses, 1):
        if not all([e.get("date"), e.get("category"), e.get("amount"), e.get("description")]):
            print(f"Entry {i}: Incomplete - skipping")
            continue
        print(f"{i}. Date: {e['date']} | Category: {e['category']} | Amount: Rs.{e['amount']} | Desc: {e['description']}")

def set_budget():
    global monthly_budget
    try:
        monthly_budget = float(input("Enter total amount for monthly budget: "))
        if monthly_budget < 0:
            print("Budget cannot be negative")
            monthly_budget = 0
        else:
            print(f"Monthly budget set to Rs.{monthly_budget}")
    except:
        print("Error: Invalid amount")

def track_budget():
    global monthly_budget
    if monthly_budget == 0:
        print("No budget set. Use option 3 to set a monthly budget first.")
        return
    
    current_month = datetime.now().strftime("%Y-%m")
    month_expenses = [e for e in expenses if e["date"].startswith(current_month)]
    total_spent = sum(e["amount"] for e in month_expenses)
    remaining = monthly_budget - total_spent
    
    print(f"\n--- Budget Track for {current_month} ---")
    print(f"Monthly Budget: Rs.{monthly_budget}")
    print(f"Total spent so far: Rs.{total_spent}")
    
    if total_spent > monthly_budget:
        print("You have exceeded your budget!")
    else:
        print(f"You have Rs.{remaining} left for the month")

def menu():
    load_expenses()
    while True:
        print("\n--- Personal Expense Tracker ---")
        print("1. Add expense")
        print("2. View expenses")
        print("3. Set monthly budget")
        print("4. Track budget")
        print("5. Save expenses")
        print("6. Exit")
        
        choice = input("Choose option 1-6: ")
        
        if choice == "1":
            add_expense()
        elif choice == "2":
            view_expenses()
        elif choice == "3":
            set_budget()
        elif choice == "4":
            track_budget()
        elif choice == "5":
            save_expenses()
        elif choice == "6":
            save_expenses()
            print("Exiting. Goodbye!")
            break
        else:
            print("Invalid choice. Enter 1-6")

if __name__ == "__main__":
    menu()
```
