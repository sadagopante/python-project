```python
import json
import os
import hashlib
from getpass import getpass

USERS_FILE = "users.json"
TASKS_FILE = "tasks.json"

def load_data(file):
    if not os.path.exists(file):
        return {}
    with open(file, "r") as f:
        try:
            return json.load(f)
        except:
            return {}

def save_data(file, data):
    with open(file, "w") as f:
        json.dump(data, f, indent=4)

def hash_password(password):
    return hashlib.sha256(password.encode()).hexdigest()

def register():
    users = load_data(USERS_FILE)
    username = input("Enter new username: ").strip()
    if not username:
        print("Username cannot be empty")
        return None
    if username in users:
        print("Username already exists")
        return None

    password = getpass("Enter password: ")
    if len(password) < 4:
        print("Password must be at least 4 characters")
        return None

    users[username] = hash_password(password)
    save_data(USERS_FILE, users)

    # Initialize empty task list for user
    tasks = load_data(TASKS_FILE)
    tasks[username] = []
    save_data(TASKS_FILE, tasks)

    print("Registration successful!")
    return username

def login():
    users = load_data(USERS_FILE)
    username = input("Username: ").strip()
    password = getpass("Password: ")

    if username in users and users[username] == hash_password(password):
        print(f"Welcome, {username}!")
        return username
    else:
        print("Invalid username or password")
        return None

def get_user_tasks(username):
    tasks = load_data(TASKS_FILE)
    return tasks.get(username, [])

def save_user_tasks(username, user_tasks):
    tasks = load_data(TASKS_FILE)
    tasks[username] = user_tasks
    save_data(TASKS_FILE, tasks)

def add_task(username):
    title = input("Task title: ").strip()
    if not title:
        print("Title cannot be empty")
        return

    desc = input("Description: ").strip()
    user_tasks = get_user_tasks(username)

    task = {
        "id": len(user_tasks) + 1,
        "title": title,
        "description": desc,
        "completed": False
    }
    user_tasks.append(task)
    save_user_tasks(username, user_tasks)
    print("Task added!")

def view_tasks(username):
    user_tasks = get_user_tasks(username)
    if not user_tasks:
        print("No tasks found.")
        return

    print("\n--- Your Tasks ---")
    for t in user_tasks:
        status = "✓ Done" if t["completed"] else "✗ Pending"
        print(f"ID: {t['id']} | {t['title']} | {status}")
        print(f" Desc: {t['description']}\n")

def mark_completed(username):
    user_tasks = get_user_tasks(username)
    if not user_tasks:
        print("No tasks to mark.")
        return

    view_tasks(username)
    try:
        tid = int(input("Enter Task ID to mark completed: "))
    except:
        print("Invalid ID")
        return

    for t in user_tasks:
        if t["id"] == tid:
            t["completed"] = True
            save_user_tasks(username, user_tasks)
            print("Task marked as completed!")
            return
    print("Task ID not found")

def delete_task(username):
    user_tasks = get_user_tasks(username)
    if not user_tasks:
        print("No tasks to delete.")
        return

    view_tasks(username)
    try:
        tid = int(input("Enter Task ID to delete: "))
    except:
        print("Invalid ID")
        return

    new_tasks = [t for t in user_tasks if t["id"]!= tid]
    if len(new_tasks) == len(user_tasks):
        print("Task ID not found")
    else:
        # Re-number IDs to keep them sequential
        for i, t in enumerate(new_tasks, 1):
            t["id"] = i
        save_user_tasks(username, new_tasks)
        print("Task deleted!")

def task_menu(username):
    while True:
        print(f"\n--- Task Manager | User: {username} ---")
        print("1. Add task")
        print("2. View tasks")
        print("3. Mark task as completed")
        print("4. Delete task")
        print("5. Logout")

        choice = input("Choose 1-5: ")

        if choice == "1":
            add_task(username)
        elif choice == "2":
            view_tasks(username)
        elif choice == "3":
            mark_completed(username)
        elif choice == "4":
            delete_task(username)
        elif choice == "5":
            print("Logged out.")
            break
        else:
            print("Invalid choice")

def main():
    while True:
        print("\n=== Task Manager ===")
        print("1. Login")
        print("2. Register")
        print("3. Exit")

        choice = input("Choose 1-3: ")

        if choice == "1":
            user = login()
            if user:
                task_menu(user)
        elif choice == "2":
            user = register()
            if user:
                task_menu(user)
        elif choice == "3":
            print("Goodbye!")
            break
        else:
            print("Invalid choice")

if __name__ == "__main__":
    main()
```
